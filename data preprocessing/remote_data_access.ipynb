```python
import pandas as pd
import numpy as np
```

<b>accessing the MONGO_URI url from configuration file to connect with cluster on mongodb atlas </b>

```python
from dotenv import load_dotenv
import os

load_dotenv('../config/credentials.env')
    
MONGO_URI = os.getenv("MONGO_URI")
print(MONGO_URI)
```

<b>accessing the collection on cluster </b>

```python
from pymongo import MongoClient

client = MongoClient(MONGO_URI, tls=True, tlsAllowInvalidCertificates=True)
db = client["ecommerce_db"]                 
collection = db["sales"]  
```

<b>converting the collection into dataframe </b>

```python
documents = list(collection.find())
df= pd.DataFrame(documents)
```

<b>saving the dataframe into csv on folder for further access</b>

```python
df.to_csv("raw_data.csv")
```
